# PEEK ResNet50 Demo


Rendered output images are written to `figures/resnet50_demo/`.
PEEK mean/variance and RVC tables are written to `metrics/resnet50_demo/`.


```python
import torch
from peek.extractors.torch import extract_torch_latents
from peek import compute_feature_folder_metrics, plot_PEEK

extract_torch_latents(
    model="resnet50",
    images_glob="image_samples/imagenet_valset/*.JPEG",
    out_dir="feature_maps/resnet50_demo",
    img_size=224,
    modules=None,
    device="cuda" if torch.cuda.is_available() else "cpu",
    capture_mode="conv_post_activation",
    verbose=True,
)

plot_PEEK(
    modules=[0, 3, 12, 24, 42],
    image_dir="image_samples/imagenet_valset",
    feature_folder="feature_maps/resnet50_demo",
    save_path="figures/resnet50_demo",
    run_path=False,
    verbose=True,
)

compute_feature_folder_metrics(
    feature_folder="feature_maps/resnet50_demo",
    modules=None,
    out_csv="metrics/resnet50_demo/per_image_metrics.csv",
    summary_csv="metrics/resnet50_demo/module_summary.csv",
    verbose=True,
)
```
