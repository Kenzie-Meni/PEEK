# PEEK ConvNeXt Demo


Rendered output images are written to `figures/convnext_demo/`.
PEEK mean/variance and RVC tables are written to `metrics/convnext_demo/`.
ConvNeXt uses spatial `CNBlock` outputs for this torch metrics path.


```python
import torch
from peek.extractors.torch import extract_torch_latents
from peek import compute_feature_folder_metrics, plot_PEEK

extract_torch_latents(
    model="convnext_base",
    images_glob="image_samples/imagenet_valset/*.JPEG",
    out_dir="feature_maps/convnext_demo",
    img_size=224,
    modules=None,
    device="cuda" if torch.cuda.is_available() else "cpu",
    capture_mode="conv_post_activation",
    verbose=True,
)

plot_PEEK(
    modules=[0, 5, 14, 27, 35],
    image_dir="image_samples/imagenet_valset",
    feature_folder="feature_maps/convnext_demo",
    save_path="figures/convnext_demo",
    run_path=False,
    verbose=True,
)

compute_feature_folder_metrics(
    feature_folder="feature_maps/convnext_demo",
    modules=None,
    out_csv="metrics/convnext_demo/per_image_metrics.csv",
    summary_csv="metrics/convnext_demo/module_summary.csv",
    verbose=True,
)
```
