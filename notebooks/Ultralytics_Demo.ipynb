# YOLO 26 Example

Rendered YOLO26 output images are written to `figures/imagenet_valset_y26s/`.
Prediction overlays are written to `runs/peek_detect/imagenet_valset/`.


```python
import torch
from peek.extractors.ultralytics import extract_ultralytics_latents
from peek import plot_PEEK

extract_ultralytics_latents(
    weights="weights/yolo26s.pt",
    image_dir="image_samples/imagenet_valset",
    out_dir="feature_maps/imagenet_valset_y26s",
    imgsz=640,
    modules=[1, 7, 15, 18, 21],
    verbose=True,
    limit=1,
)

plot_PEEK(
    modules=[1, 7, 15, 18, 21],
    image_dir="image_samples/imagenet_valset",
    feature_folder="feature_maps/imagenet_valset_y26s",
    save_path="figures/imagenet_valset_y26s",
    run_path="auto",
    weights_path="weights/yolo26s.pt",
    imgsz=640,
    verbose=True,
    limit=1,
)

plot_PEEK(
    modules=[1, 7, 15, 18, 21],
    image_dir="image_samples/imagenet_valset",
    feature_folder="feature_maps/imagenet_valset_y26s",
    save_path=False,
    run_path="runs/peek_detect/imagenet_valset",
    verbose=True,
    limit=1,
)
```
