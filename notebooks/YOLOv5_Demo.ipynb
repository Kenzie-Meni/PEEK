## Run Inference and Save Feature Maps

(Note: for large datasets, this will create tens of GB of data. Please ensure you have sufficient hard drive space.)

Rendered output images are written to `figures/GSFC_satellite/`.
Prediction overlays used by the notebook are written to `runs/peek_detect/GSFC_satellite/`.
PEEK mean/variance and RVC tables are written to `metrics/GSFC_satellite/`.


```python
from peek.extractors.yolov5 import extract_yolov5_latents
from peek import compute_feature_folder_metrics, plot_PEEK

extract_yolov5_latents(
    weights_path="weights/yolov5-satellite-components-det-baseline.pt",
    images_glob="image_samples/GSFC_satellite/*",
    out_dir="feature_maps/GSFC_satellite",
    imgsz=640,
    modules=None,
)

plot_PEEK(
    modules=[1, 7, 16, 19, 22],
    image_dir="image_samples/GSFC_satellite",
    feature_folder="feature_maps/GSFC_satellite",
    save_path="figures/GSFC_satellite",
    run_path="auto",
    weights_path="weights/yolov5-satellite-components-det-baseline.pt",
    verbose=True,
    limit=1,
)

plot_PEEK(
    modules=[1, 7, 16, 19, 22],
    image_dir="image_samples/GSFC_satellite",
    feature_folder="feature_maps/GSFC_satellite",
    save_path=False,
    run_path="runs/peek_detect/GSFC_satellite",
    verbose=True,
    limit=1,
)

compute_feature_folder_metrics(
    feature_folder="feature_maps/GSFC_satellite",
    modules=None,
    out_csv="metrics/GSFC_satellite/per_image_metrics.csv",
    summary_csv="metrics/GSFC_satellite/module_summary.csv",
    verbose=True,
)
```
